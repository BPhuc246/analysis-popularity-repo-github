```python
import json
from pathlib import Path
import pandas as pd

ORIGINAL = Path("../data/raw/github_repos_complete.csv")
FILLED = Path("output/topics_filled.csv")
OUTPUT = Path("data/raw/github_repos_complete_topics_filled.csv")
ONLY_CONFIDENT = False   # True: chỉ điền repo không bị đánh dấu low_confidence

base = pd.read_csv(ORIGINAL, dtype=str, keep_default_na=False, encoding="utf-8-sig")
fill = pd.read_csv(FILLED, dtype=str, keep_default_na=False, encoding="utf-8-sig")

if ONLY_CONFIDENT:
    fill = fill[fill["low_confidence"].str.upper() != "TRUE"]

fill_map = dict(zip(fill["full_name"].str.strip(), fill["topics_filled"]))

is_empty = base["topics"].str.strip().isin(["", "[]"])
targets = base["full_name"].str.strip().map(fill_map)
replace = is_empty & targets.notna() & (targets != "[]")

base.loc[replace, "topics"] = targets[replace]
base.to_csv(OUTPUT, index=False, encoding="utf-8-sig")

print(f"Đã điền {int(replace.sum())} / {int(is_empty.sum())} repo có topics rỗng")
print(f"Lưu tại {OUTPUT}")
```
